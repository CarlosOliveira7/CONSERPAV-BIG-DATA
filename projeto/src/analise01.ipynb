```python
import pandas as pd

df = pd.read_csv("../data/base_de_dados.csv", sep=",", decimal=".", encoding="latin-1")
df.columns = df.columns.str.strip()

print("Colunas disponíveis:", df.columns.tolist())
print(f"Total de registros: {len(df)} | Atributos: {df.shape[1]}")
print("Percentual de nulos por coluna:")
print((df.isna().mean() * 100).round(2))
print("Duplicidades:", df.duplicated().sum())

coluna_remuneracao = "valor_remuneracao_media"
if coluna_remuneracao in df.columns:
    print(df[coluna_remuneracao].describe())        
    print((df[coluna_remuneracao] == 0).sum())      
else:
    print(f"A coluna '{coluna_remuneracao}' não foi encontrada.")

if "idade" in df.columns:
    print(df["idade"].describe())
else:
    print("A coluna 'idade' não foi encontrada.")
```
